WSL o también conocido como Subsistema de Windows para Linux, es una característica de Windows que permite ejecutar un entorno de Linux dentro de Windows sin necesidad de una máquina virtual independiente o arranque dual. ### Conda

Conda es un sistema de gestión de paquetes multiplataforma de código abierto. Inicialmente se creó para resolver los problemas de gestión de paquetes. Conda también es un gestor de entornos, esto es de gran ayuda en el supuesto de que necesitas correr un software que necesita una versión diferente de Python a la que ya tienes instalada por ejemplo. En ese caso se puede configurar un entorno y cambiar fácilmente de uno a otro para poder utilizar versiones diferentes de Python u otra herramienta en el mismo dispositivo. 

![Imagen1](Imagen1.png)

Micromamba forma parte del ecosistema Mamba y está diseñado para ofrecer una alternativa minimalista al gestor de paquetes completo de Conda, ya que se trata de una versión mucho más ligera y rápida. Asimismo, Micromamba admite todas las funciones principales de Conda, incluyendo la creación de entorno, resolución de dependencias e instalación de paquetes. ¿Qué es un paquete? Un paquete, tanto de Micromamba como de Conda es un archivo comprimido que contiene todos los elementos necesarios para ejecutar un software o librería específica.


### ¿Para qué sirve cada herramienta?

#### Fastp

Fastp es un preprocesador de lecturas FASTQ escrito en C++ y multihilo. Reemplaza en una sola pasada lo que antes requería encadenar FastQC + Trimmomatic/Cutadapt + FastQC. Se utiliza para:

-   Recorte de adaptadores
-	Filtrado por calidad
-	Recorte por calidad
-	Filtrado por longitud
-	Corrección de bases
-	Poly-G trimming
-	Deduplicación
-	Reportes de calidad antes/después en HTML y JSON

Lo usamos porque adaptadores y bases de baja calidad reducen la tasa de mapeo e introducen falsos positivos en la cuantificación y en la llamada de variantes. Un comando ejemplo puede ser:

```python
fastp \
  -i muestra_R1.fastq.gz -I muestra_R2.fastq.gz \
  -o limpio_R1.fastq.gz  -O limpio_R2.fastq.gz \
  --detect_adapter_for_pe \
  --qualified_quality_phred 20 \
  --length_required 36 \
  --trim_poly_g \
  --thread 8 \
  --html muestra_fastp.html --json muestra_fastp.json
```


#### Subread

Subread a diferencia de fastp no es solo un programa, sino que es una suite de herramientas en C. Al instalar el paquete se obtienen varios ejecutables:

| Ejecutable | Función |
| :--- | :--- |
| `subread-buildindex` | construye el índice del genoma de referencia |
| `subread-align` | alineador de propósito general (DNA-seq y RNA-seq), basado en el algoritmo *seed-and-vote* |
| `subjunc` | alineador de RNA-seq enfocado en **detectar uniones exón-exón** (splicing) |
| `featureCounts` | **contador de lecturas por feature** — la herramienta más usada de la suite |
| `exactSNP` | llamado de SNPs |
| `flattenGTF`, `repair` | utilidades auxiliares |

Subread maneja correctamente los casos difíciles como las lecturas paired-end, lecturas multi-mapeadas, lecturas que solapan varios features y datos stranded. Su salida es simple y estable. Además viene junto con un resumen que MultiQC lee directamente. Un comando ejemplo puede ser:

```python
# Índice del genoma (una sola vez por referencia)
subread-buildindex -o indice/genoma_idx genoma.fa

# Alineamiento de RNA-seq consciente de splicing
subjunc -i indice/genoma_idx -r limpio_R1.fastq.gz -R limpio_R2.fastq.gz \
        -T 8 -o muestra.bam

# Conteo por gen
featureCounts -a anotacion.gtf -o conteos.txt \
              -t exon -g gene_id \
              -p --countReadPairs \
              -s 2 -T 8 \
              muestra1.bam muestra2.bam muestra3.bam
```


#### MultiQC

Se trata de una herramienta en Python que recorre un directorio, reconoce automáticamente los archivos log y reporte de más de 100 herramientas bioinformáticas y los fusiona en un único informe HTML interactivo. Se utiliza para:

-	Comparar todas las muestras a la vez
-	Detectar valores atípicos
-	Documentar el proyecto

MultiQC es muy relevante en el paso de verificación del flujo; ya que los problemas de un experimento de secuenciación casi nunca se ven mirando una muestra aislada, sino que se ven a, comparar todas juntas. MultiQC convierte esta comparación en algo inmediato y hace que el QC quede documentado en vez de ser una impresión subjetiva. Un comando ejemplo puede ser:

```python
# Rastrea el directorio actual y todos sus subdirectorios
multiqc .

# Con directorio de salida y nombre de reporte propios
multiqc resultados/ -o reporte_multiqc/ -n reporte_proyecto.html --title "Proyecto RNA-seq"

# Sobrescribir un reporte previo
multiqc . -f
```


#### Bismark

Bismark ha sido instalado debido a que vamos a trabajar con muestras de rnaseq pero también con muestras tratadas con bisulfito. La secuenciación con bisulfito convierte el estado de metilación de cada citosina en un cambio de base: la citosina no metilada se lee como timina y la citosina metilada sigue siendo citosina. 

![Imagen2](Imagen2.png)

Esa conversión hace fallar a los alineadores comunes. En este caso no sería lo más acertado utilizar Subread con muestras tratadas con bisulfito. Bismark mapea las lecturas tratadas con bisulfito contra una referencia y, a continuación, extrae las llamadas de metilación directamente de los alineamientos, manteniendo separados los contextos CpG, CHG y CHH. 

#### SRAToolkit

SRAToolkit sirve para descargar y extraer datos públicos de secuenciación genética desde la base de datos SRA del NCBI. Es necesario instalar este programa para descargar archivos de más de 5 giga bases.
